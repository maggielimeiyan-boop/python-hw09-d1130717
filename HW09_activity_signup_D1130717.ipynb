```python
測試一：原始資料
```

```python
# --- 測試一：原始資料 ---
# 固定的兩個活動名稱，以 tuple 保存
activity_names = ("桌遊體驗", "電影欣賞")

# 收到的原始文字名單，保留原值
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "

# 用活動名稱查詢每人費用
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

# 1. 讀取活動名稱，建立兩份整理後的串列（保留順序與重複項目）
board_name, movie_name = activity_names
board_list = board_raw.replace(" ", "").lower().split(",")
movie_list = movie_raw.replace(" ", "").lower().split(",")

# 2. 使用集合去重並比較名單
board_set = set(board_list)
movie_set = set(movie_list)

common_participants = board_set & movie_set       # 交集：共同參加者
board_only = board_set - movie_set                # 差集：只參加桌遊者
movie_only = movie_set - board_set                # 差集：只參加電影者
all_participants = board_set | movie_set          # 聯集：全部參加者

# 3. 建立排序序列，使用切片取出前兩位作為通知名單預覽
all_sorted = sorted(all_participants)
notify_preview = all_sorted[:2]

# 4. 使用字典保存摘要，計算人數與應收費用
board_fee = unit_fees[board_name]
movie_fee = unit_fees[movie_name]

summary = {
    "board_count": len(board_set),
    "movie_count": len(movie_set),
    "common_count": len(common_participants),
    "total_unique_count": len(all_participants),
    "total_registrations": len(board_set) + len(movie_set),
    "board_revenue": len(board_set) * board_fee,
    "movie_revenue": len(movie_set) * movie_fee,
}
summary["total_revenue"] = summary["board_revenue"] + summary["movie_revenue"]

# 5. 使用 f-string 輸出完整摘要
print("HW09 活動報名摘要（測試一）")
print(f"活動：{board_name}，每人 {board_fee:.2f} 元")
print(f"活動：{movie_name}，每人 {movie_fee:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{sorted(board_set)}")
print(f"電影去重名單：{sorted(movie_set)}")
print(f"桌遊人數：{summary['board_count']} 人")
print(f"電影人數：{summary['movie_count']} 人")
print(f"共同參加者：{sorted(common_participants)}")
print(f"共同人數：{summary['common_count']} 人")
print(f"只參加桌遊者：{sorted(board_only)}")
print(f"只參加電影者：{sorted(movie_only)}")
print(f"全部參加者：{all_sorted}")
print(f"總不重複人數：{summary['total_unique_count']} 人")
print(f"報名人次：{summary['total_registrations']} 人次")
print(f"通知名單前兩位：{notify_preview}")
print(f"桌遊應收：{summary['board_revenue']:.2f} 元")
print(f"電影應收：{summary['movie_revenue']:.2f} 元")
print(f"總應收：{summary['total_revenue']:.2f} 元")
```
